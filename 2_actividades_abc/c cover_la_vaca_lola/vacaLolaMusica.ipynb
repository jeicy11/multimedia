```python
# video original
archivo = '/content/drive/MyDrive/Multimedia2026/vaca/baile.mp4'
```

```python
import cv2
```

```python
video = cv2.VideoCapture(archivo)  # abrir el video
```

```python
# PARA VERIFICAR
print(video.isOpened())  # verificar si el video se abrió correctamente
```

```python
# extraer frames del video y guardarlos como imágenes
contador = 1
while True:
    retorno, frame = video.read()  # leer un frame
    if not retorno:  # si no hay más frames, salir del bucle
        break
    cv2.imwrite('/content/drive/MyDrive/Multimedia2026/vaca/frames/imagen'+str(contador)+'.jpg', frame)  # guardar frame
    contador += 1
```

```python
import os
import re

directorio = '/content/drive/MyDrive/Multimedia2026/vaca/frames/'

imagenes = sorted(
    os.listdir(directorio),
    key=lambda x: int(re.findall(r'\d+', x)[0])
)

print(imagenes[:15])
```

```python
# leer la primera imagen para obtener sus dimensiones
frame = cv2.imread(directorio + imagenes[0])
alto, ancho, _ = frame.shape
```

```python
# configurar el codec para el video de salida
driver = cv2.VideoWriter_fourcc(*'mp4v')
# crear video de salida con las dimensiones de las imágenes
video = cv2.VideoWriter('/content/drive/MyDrive/Multimedia2026/vaca/video.mp4', driver, 40, (ancho, alto))
```

```python
# recorrer todas las imágenes para escribirlas en el video
for imagen in imagenes:
    archivo = '/content/drive/MyDrive/Multimedia2026/vaca/frames/' + imagen
    frame = cv2.imread(archivo)
    # escribir el frame en el video
    video.write(frame)
# cerrar el video
video.release()
```
